```python
import pandas as pd
import json
from statistics import median
```

```python
df = pd.read_csv("Project_havePolicy.csv")

# Count the number of projects for each reporting mechanism
reporting_counts = df['SecurityPolicy_reporting_mechanism'].value_counts(dropna=False)
total_projects = reporting_counts.sum()

reporting_summary = reporting_counts.reset_index()
reporting_summary.columns = ['Reporting Mechanism', 'Project Count']
reporting_summary['Percentage'] = (reporting_summary['Project Count'] / total_projects) * 100

display(df)
print("\nNumber of projects for each Reporting mechanism:")
for _, row in reporting_summary.iterrows():
    mechanism = row['Reporting Mechanism']
    count = row['Project Count']
    percent = row['Percentage']
    print(f"- {mechanism}: {count} projects ({percent:.2f}%)")

print(f"\nTotal projects: {total_projects}")
```
